### Dataset Grain Conclusion

The dataset is stored at the **order-item level**, meaning that one row represents one product/item within an order rather than one unique order.

The dataset contains 3,266 rows but only 3,153 unique order IDs. A total of 103 orders appear across multiple rows.

Investigation of repeated order IDs confirms that these rows are not duplicate records. Instead, they represent different products or product variations purchased within the same order.

Therefore:

- `No. Pesanan` is not unique at the row level.
- One order may contain multiple products or variations.
- Order-level KPIs must be calculated using unique order IDs to prevent double counting.
- Product and item-level metrics can be calculated using individual rows.

### Results of the additional investigation

First, the 120 `payment_at` entries that converted to `NaT` turned out not to be parsing errors; the raw values ​​were simply "-". All 120 of these rows were marked as "Cancelled." This indicates that these orders had already entered the fulfillment or logistics stage before ultimately being cancelled.

### Processing Category

So, we use the percentile as the operational monitoring threshold:

### Processing Time Classification

Processing-time categories are defined using the empirical distribution of the dataset:

- Normal: ≤ 90th percentile
- Attention: > P90 to P95
- Slow: > P95 to P99
- Critical: > P99

These categories are intended for operational monitoring and prioritization rather than representing predefined service-level agreements.

```python
p90_processing = df_clean[
    "processing_time_minutes"
].quantile(0.90)

p95_processing = df_clean[
    "processing_time_minutes"
].quantile(0.95)

p99_processing = df_clean[
    "processing_time_minutes"
].quantile(0.99)

print("P90:", p90_processing)
print("P95:", p95_processing)
print("P99:", p99_processing)
```

```python
def classify_processing_time(x):
    if pd.isna(x):
        return "Not Processed"
    elif x <= p90_processing:
        return "Normal"
    elif x <= p95_processing:
        return "Attention"
    elif x <= p99_processing:
        return "Slow"
    else:
        return "Critical"

df_clean["processing_category"] = (
    df_clean["processing_time_minutes"]
    .apply(classify_processing_time)
)
```

## Order-Level Operational Table

The raw dataset is stored at the order-item level. A separate order-level table is therefore created to calculate operational KPIs without double-counting multi-item orders.

Order-level operational fields are aggregated carefully, while item-level return information is summarized across all items within each order.

```python
df_orders = (
    df_clean
    .groupby("order_id", as_index=False)
    .agg(
        order_status=("order_status", "first"),
        order_created_at=("order_created_at", "first"),
        order_date=("order_date", "first"),
        payment_at=("payment_at", "first"),
        shipping_deadline=("shipping_deadline", "first"),
        shipping_arranged_at=("shipping_arranged_at", "first"),
        order_completed_at=("order_completed_at", "first"),

        payment_lead_time_minutes=(
            "payment_lead_time_minutes",
            "first"
        ),

        processing_time_minutes=(
            "processing_time_minutes",
            "first"
        ),

        completion_time_hours=(
            "completion_time_hours",
            "first"
        ),

        deadline_margin_minutes=(
            "deadline_margin_minutes",
            "first"
        ),

        is_cancelled=("is_cancelled", "max"),

        has_started_fulfillment=(
            "has_started_fulfillment",
            "max"
        ),

        cancellation_stage=(
            "cancellation_stage",
            "first"
        ),

        processing_category=(
            "processing_category",
            "first"
        ),

        is_late_shipment=(
            "is_late_shipment",
            "max"
        ),

        is_returned=(
            "is_returned",
            "max"
        ),

        returned_items=(
            "returned_quantity",
            "sum"
        ),

        total_items=(
            "quantity",
            "sum"
        ),

        unique_products=(
            "product_name",
            "nunique"
        )
    )
)
```

### Cancellation Reason Standardization

Currently, the cancellation reason appears as a long sentence from Shopee. That is not ideal for the dashboard.

```python
def categorize_cancellation_reason(reason):

    if pd.isna(reason):
        return "Not Cancelled"

    reason_lower = reason.lower()

    if "belum dibayar" in reason_lower:
        return "Unpaid Order"

    elif "pengiriman gagal" in reason_lower:
        return "Delivery Failed"

    elif "paket hilang" in reason_lower:
        return "Package Lost"

    elif "alamat" in reason_lower or "delivery address" in reason_lower:
        return "Change Address"

    elif "ubah pesanan" in reason_lower or "mengubah pesanan" in reason_lower:
        return "Change Order"

    elif "voucher" in reason_lower:
        return "Voucher Issue"

    elif "tidak responsif" in reason_lower:
        return "Seller Responsiveness"

    elif "pembayaran sulit" in reason_lower:
        return "Payment Difficulty"

    elif "tidak ingin membeli" in reason_lower:
        return "Changed Mind"

    elif "lebih murah" in reason_lower:
        return "Found Cheaper Alternative"

    elif "lainnya" in reason_lower:
        return "Other"
    elif ("penjual tidak mengatur pengiriman tepat waktu" in reason_lower
        or "penjual gagal mengirimkan pesanan tepat waktu" in reason_lower):
        return "Seller Late Shipment"

    elif "lainnya" in reason_lower:
        return "Other"

    else:
        return "Other"
```

```python
df_clean["cancellation_category"] = (
    df_clean["cancellation_reason"]
    .apply(categorize_cancellation_reason)
)
```

## Key Operational Insights

### 1. Order Completion Performance
A total of 3,153 unique orders were recorded during October 2025. Of these, 2,650 orders were completed and 503 were cancelled, resulting in an 84.05% completion rate and a 15.95% cancellation rate.

### 2. Fulfillment Processing Performance
Among 2,712 orders that entered fulfillment, the median processing time was approximately 18.88 hours, while 90% were processed within approximately 32.10 hours.

### 3. Critical Processing Cases Were Limited
26 processed orders, approximately 0.96%, were classified as Critical based on processing times above the 99th-percentile monitoring threshold.

### 4. Shipping Deadline Compliance Was High
Only 9 processed orders were arranged for shipping after their recorded marketplace deadline, resulting in a late-shipment rate of approximately 0.33% and an on-time shipping-arrangement rate of 99.67%.

### 5. Most Cancellations Occurred Before Fulfillment
Of 503 cancelled orders, 441 (87.67%) occurred before fulfillment began, while 62 (12.33%) occurred after fulfillment had already been initiated. Post-fulfillment cancellations may require additional operational attention because fulfillment or logistics activity has already started.

### 6. Return Incidence Was Relatively Limited
36 completed orders contained at least one returned item, corresponding to a return-order rate of approximately 1.36%. A total of 36 out of 3,513 ordered items were returned, representing an item return rate of approximately 1.02%.

### 7. Daily Order Volume Was Not Strongly Associated With Processing Time
The correlation between daily order volume and median processing time was approximately 0.076, indicating little linear association between workload volume and processing time during the observed period.
